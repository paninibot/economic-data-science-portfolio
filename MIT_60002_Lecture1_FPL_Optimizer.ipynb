```python
# Cell 1: Player Class & Mock Dataset
class Player:
    def __init__(self, name, position, cost, points):
        self.name = name
        self.position = position  # 'GKP', 'DEF', 'MID', 'FWD'
        self.cost = cost          # in millions (£)
        self.points = points      # expected total points

    def value_density(self):
        """Points per million spent - our greedy metric"""
        return self.points / self.cost

# Sample pool of Premier League players
PLAYER_POOL = [
    # Forwards
    Player("Haaland", "FWD", 14.0, 240),
    Player("Isak", "FWD", 8.5, 160),
    Player("Watkins", "FWD", 9.0, 175),
    Player("Joao Pedro", "FWD", 5.5, 110),
    Player("Vardy", "FWD", 5.0, 95),

    # Midfielders
    Player("Palmer", "MID", 10.5, 210),
    Player("Saka", "MID", 10.0, 200),
    Player("Salah", "MID", 13.0, 230),
    Player("Son", "MID", 9.5, 180),
    Player("Bowen", "MID", 7.5, 145),
    Player("Gordon", "MID", 7.0, 135),
    Player("Rogers", "MID", 5.5, 105),
    Player("Diallo", "MID", 5.0, 90),

    # Defenders
    Player("Gabriel", "DEF", 6.0, 140),
    Player("Saliba", "DEF", 6.0, 135),
    Player("Virgil", "DEF", 6.5, 130),
    Player("Pedro Porro", "DEF", 5.5, 115),
    Player("Konsa", "DEF", 4.5, 95),
    Player("Burn", "DEF", 4.5, 90),
    Player("Mykolenko", "DEF", 4.0, 80),

    # Goalkeepers
    Player("Raya", "GKP", 5.5, 130),
    Player("Pickford", "GKP", 5.0, 115),
    Player("Fabianski", "GKP", 4.0, 70),
]
print(f"Loaded {len(PLAYER_POOL)} players into the pool successfully!")
```

```python
# Cell 2: The Greedy Algorithm & Execution
def greedy_fpl_selector(pool, max_budget=75.0, max_players=8):
    """
    Greedy Strategy:
    1. Sort players by 'value density' (Points / Cost) descending.
    2. Pick players one by one if they fit within the remaining budget and squad limit.
    """
    # Sort by points per million (descending)
    sorted_pool = sorted(pool, key=lambda p: p.value_density(), reverse=True)

    selected_squad = []
    total_cost = 0.0
    total_points = 0

    print("--- EVALUATING PLAYERS VIA GREEDY HEURISTIC ---")
    for player in sorted_pool:
        if len(selected_squad) < max_players and (total_cost + player.cost) <= max_budget:
            selected_squad.append(player)
            total_cost += player.cost
            total_points += player.points
            print(f"Picked: {player.name:<12} | Cost: £{player.cost}m | Points: {player.points} | Density: {player.value_density():.2f}")

    return selected_squad, total_cost, total_points

# Run the optimizer with an £75m budget cap for 8 players
budget = 75.0
squad, cost, points = greedy_fpl_selector(PLAYER_POOL, max_budget=budget, max_players=8)

print("\n" + "="*45)
print(f"FINAL GREEDY SQUAD RESULTS:")
print(f"Total Squad Cost: £{cost:.1f}m / £{budget}m budget")
print(f"Total Expected Points: {points}")
print("="*45)
```
